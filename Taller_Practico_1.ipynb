## 2.2 Variables cuantitativas

### Variable cuantitativa 1: EDAD

```python
k = 7
d = 0

df['Intervalos_edad'] = pd.cut(df['EDAD'], bins = k, include_lowest = True, precision = d)
frecuencia = df.groupby('Intervalos_edad', observed = True).size()

tabla_edad = pd.DataFrame({
    'fi': frecuencia,
    'hi': (frecuencia/n)*100,
    'Fi': frecuencia.cumsum(),
    'Hi': ((frecuencia/n)*100).cumsum()
})
round(tabla_edad, 1)
```

Se utiliza un **histograma**, ya que la variable es cuantitativa continua y sus datos están agrupados en intervalos.

```python
categorias = tabla_edad.index
f = tabla_edad['hi']

fig, ax = plt.subplots(figsize = (7,5))
histograma = ax.bar(categorias.astype(str), f, color = 'steelblue', edgecolor = 'black', width = 1)

ax.set_title('Distribución de estudiantes según edad')
ax.set_xlabel('Edad (años)')
ax.set_ylabel('Porcentaje de estudiantes (%)')
ax.bar_label(histograma, fmt = '%.1f%%')
plt.xticks(rotation = 45, ha = 'right')
plt.show()
```

**Interpretación de la tabla y el gráfico:**
*  $f_1$: 4.171 estudiantes tienen entre 18 y 25 años.
*  $h_1$: el 72,0% de los estudiantes tiene entre 18 y 25 años (mayor frecuencia).
*  $F_2$: 5.155 estudiantes tienen entre 18 y 32 años.
*  $H_3$: el 96,2% de los estudiantes tiene entre 18 y 39 años.
*  $h_7$: solo el 0,1% de los estudiantes tiene más de 60 y hasta 67 años (menor frecuencia).

El histograma presenta una distribución **asimétrica positiva** (cola hacia la derecha): la mayoría de los estudiantes es joven y la frecuencia disminuye a medida que aumenta la edad.

```python
media_edad = df['EDAD'].mean()
mediana_edad = df['EDAD'].median()
moda_edad = df['EDAD'].mode()

print(f'Media: {media_edad:.1f}\nMediana: {mediana_edad}\nModa: {moda_edad[0]}')
```

**Medidas de tendencia central:**
*  Media: la edad promedio de los estudiantes es de 24,4 años.
*  Mediana: el 50% de los estudiantes tiene a lo más 22 años.
*  Moda: la edad más frecuente entre los estudiantes es de 19 años.

Como la moda es menor que la mediana, y esta a su vez es menor que la media, se confirma la asimetría positiva: las edades más altas elevan el promedio. Por esto, la mediana representa mejor la edad típica de los estudiantes.

```python
p10_edad = df['EDAD'].quantile(0.1)
p25_edad = df['EDAD'].quantile(0.25)
p75_edad = df['EDAD'].quantile(0.75)
p90_edad = df['EDAD'].quantile(0.9)

print(f'Percentil 10: {p10_edad}\nPercentil 25: {p25_edad}\nPercentil 75: {p75_edad}\nPercentil 90: {p90_edad}')
```

**Percentiles:**
*  Percentil 10: el 10% de los estudiantes tiene a lo más 19 años.
*  Percentil 25: el 25% de los estudiantes tiene a lo más 20 años.
*  Percentil 75: el 75% de los estudiantes tiene a lo más 26 años.
*  Percentil 90: el 90% de los estudiantes tiene a lo más 33 años; es decir, el 10% de mayor edad tiene entre 33 y 67 años.

```python
rango_edad = df['EDAD'].max() - df['EDAD'].min()
desviacion_edad = df['EDAD'].std()
cv_edad = (desviacion_edad / media_edad) * 100

print(f'Mínimo: {df["EDAD"].min()}\nMáximo: {df["EDAD"].max()}')
print(f'Rango: {rango_edad} años\nDesviación estándar: {desviacion_edad:.1f} años\nCoeficiente de variación: {cv_edad:.1f}%')
```

**Medidas de dispersión:**
*  Rango: existe una diferencia de 49 años entre el estudiante de mayor edad (67 años) y el de menor edad (18 años).
*  Desviación estándar: en promedio, las edades se alejan 6,4 años respecto de la edad promedio.
*  Coeficiente de variación: 26,1%. Es el menor de las cuatro variables cuantitativas analizadas, por lo que la edad es la variable más homogénea de este grupo.

**Aspectos que llaman la atención:** aunque el rango es amplio, el 75% de los estudiantes tiene a lo más 26 años y solo el 10% supera los 33 años. Existen estudiantes de 61, 62 y 67 años, lo que muestra que también hay adultos mayores que ingresan a la educación superior.

## a) ¿Hay áreas del conocimiento donde las carreras sean más caras? Expliquen qué criterio utilizaron para responder.

Se agrupan los estudiantes según el área del conocimiento y se calcula la **mediana del arancel anual** de cada área como criterio principal, ya que no se ve afectada por valores extremos. Como apoyo, se calcula la media, la desviación estándar y el coeficiente de variación (para evaluar si el promedio es representativo), y el **costo total estimado de la carrera**, que corresponde al arancel anual multiplicado por la duración de la carrera en años (semestres / 2).

```python

df['COSTO TOTAL ESTIMADO'] = df['VALOR ARANCEL (PESOS)'] * df['DURACION TOTAL CARRERA (SEMESTRES)'] / 2

tabla_a = df.groupby('AREA CONOCIMIENTO')['VALOR ARANCEL (PESOS)'].agg(['mean', 'median', 'std'])
tabla_a['CV (%)'] = (tabla_a['std'] / tabla_a['mean']) * 100
tabla_a['Costo total promedio'] = df.groupby('AREA CONOCIMIENTO')['COSTO TOTAL ESTIMADO'].mean()

round(tabla_a.sort_values('median', ascending = False), 1)
```

```python
tabla_a_ord = tabla_a.sort_values('median', ascending = False)
categorias = tabla_a_ord.index
f = tabla_a_ord['median']

fig, ax = plt.subplots(figsize = (8,5))
barras = ax.bar(categorias, f, color = 'coral', edgecolor = 'black')

ax.set_title('Mediana del arancel anual según área del conocimiento')
ax.set_xlabel('Área del conocimiento')
ax.set_ylabel('Mediana del arancel (pesos)')
ax.bar_label(barras, fmt = '%.0f')
plt.xticks(rotation = 45, ha = 'right')
plt.show()
```

**Respuesta:** sí, existen áreas del conocimiento donde las carreras son más caras.

*  **Derecho** es el área más cara bajo todos los criterios: el 50% de sus estudiantes paga un arancel anual de a lo más \$4.117.000 (igual a su media) y su costo total promedio es de \$28.819.000. Su coeficiente de variación es 0%, ya que corresponde a una única carrera con un único arancel.
*  **Arte y Arquitectura** (mediana \$4.091.000) y **Salud** (mediana \$3.555.000) ocupan el segundo y tercer lugar según la mediana. Salud registra el segundo mayor costo total promedio (\$17.459.643), debido a que incluye carreras largas como Medicina.
*  Las áreas más económicas son **Educación** (mediana \$1.598.000), **Agropecuaria** (\$1.600.000) y **Administración y Comercio** (\$1.942.000), que concentran una alta proporción de carreras técnicas.

Las áreas más caras superan la mediana general del arancel (\$2.234.000) y corresponden a carreras profesionales de larga duración impartidas por la universidad.
